### 3. Query Phenotype Associations with Leave-One-Lab-Out Robustness
We query a common iPSC-neuron phenotype: **reduced SYN1 puncta density**.

```python
from synaptik.store import Store
from synaptik.analyst import Analyst
from synaptik.schema import Direction

store = Store("demo.db")
analyst = Analyst(store)

result = analyst.analyze_phenotype("SYN1 puncta density", direction=Direction.DECREASE)
print(f"Total pathway associations identified: {len(result.associations)}\n")

print("--- Replicated Associations (Strong Tier) ---")
for assoc in result.strong_only():
    print(f"* Pathway: {assoc.pathway_label} | Labs: {assoc.n_labs} | Tier: {assoc.confidence_tier()}")
    for caveat in assoc.caveats():
        print(f"   ! Caveat: {caveat}")
```
